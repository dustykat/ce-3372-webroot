# Stormwater Detention Ponds

A stormwater detention pond is a constructed basin designed to temporarily store runoff during storm events and release it at a controlled rate, typically remaining dry between storms.

## Purposes

1. Mitigate Peak Discharge(s): Stormwater detention ponds are designed to temporarily store runoff during storm events and release it at a controlled rate. By attenuating peak flows, these systems help reduce downstream flooding, protect channel stability, and limit stress on existing drainage infrastructure. The outlet structure (e.g., orifice, weir, or riser) is typically configured to match pre-development or allowable discharge conditions.
2. Provide water quality benefit: In addition to flow control, detention ponds can improve water quality by promoting sedimentation of suspended solids and associated pollutants. As runoff is detained, heavier particles settle out, and some dissolved constituents may be reduced through biological uptake and chemical processes. While not as treatment-focused as retention (wet) ponds or constructed wetlands, properly designed detention systems still provide meaningful first-flush pollutant reduction and contribute to overall watershed health.

## Hydraulic Performance Estimation (Using SWMM)

Stormwater detention ponds can be represented in EPA SWMM using a storage unit node, which explicitly accounts for water volume stored as a function of depth. This allows the model to simulate both hydraulic routing (inflow/outflow) and temporary storage behavior within the basin.

### Storage Node Representation

A storage node differs from a standard junction node in that it includes volume storage within the node itself, including (if properly defined) any portion of the basin below ground surface.

Key Storage Node Attributes:

- Invert Elevation: Elevation of the bottom of the basin (reference point for all depths).
- Maximum Elevation: Elevation of the top of the storage node.
   1. Set equal to land surface if the goal is to visualize the hydraulic grade line relative to ground.
   2. Set above land surface if simulating combined surface and subsurface storage systems (e.g., detention basin with underground conveyance).
- Depth–Area Relationship: Defines how storage volume changes with depth.
  1. Tabular form – based on surveyed or designed stage–area data (preferred for real systems).
  2. Functional form – simplified geometric approximation (e.g., rectangular, trapezoidal basin).

This relationship is critical because it controls how quickly the basin fills and drains under varying inflow conditions.

Using a storage unit type node:

- A storage node explicitly considers storage in the node – including (if data are correctly supplied) the sub-grade portion of the node
- Storage Node Attributes:
  - Invert elevation (elevation of the bottom of the node)
  - Max elevation (elevation of top of node)
    1. Set to land surface to plot profile grade line in SWMM
    2. Set to land surface + added depth for simulating surface and subsurface (sewer) drainage
  - Depth-Area relationship:
    1. Tabular or
    2. Functional

### Example (Conceptualization)
Consider a detention pond receiving runoff from a 5-acre parking lot, discharging to a nearby stream. The goal is to use SWMM to approximate the hydraulic response of the system under a design storm.

Model Setup (Simplified):
- Subcatchment Area: 5 acres
- Curve Number (CN): 98 (represents nearly impervious surface)
- Rainfall Event:
  - Intensity: 1 in/hr
  - Duration: 3 hours
- Outlet Configuration:
  - Low-flow outlet: 6-inch pipe at the basin invert
  - Higher-flow outlet: 1-foot pipe located 2 feet above the basin bottom
- Flow Path: All runoff is routed through the detention pond prior to discharge to the receiving stream

This configuration creates a multi-stage outlet system, where:

- The smaller pipe controls low flows and initial drawdown
- The larger pipe activates at higher water levels, increasing discharge capacity and limiting excessive ponding

**Model Implementation Notes**

The SWMM model for this system is relatively simple to construct, consisting of:

- One subcatchment
- One storage node (detention pond)
- One or more outlet links (pipes or orifices)

**Interpretation Notes**

When reviewing results, focus on:

- Water surface elevation vs. time (does the pond overtop?)
- Outflow hydrograph (is peak discharge reduced relative to inflow?)
- Storage utilization (is the basin sized appropriately, or under/over-designed?)

This example is intended to connect conceptual design (basin + outlet structure) with modeled hydraulic performance, reinforcing how detention systems achieve peak flow mitigation in practice, and how one might employ a hydraulic model to inform design (selecting pond size to satisfy a design goal).
