## Tanglewilde Storm Sewer

## Project Context

Students are provided with an aerial image of an urban area in Houston, Texas, a scaled drawing of a storm sewer network, and land surface elevation data at selected locations. A corresponding storm sewer network topology suitable for implementation in EPA SWMM is also supplied.

The network represents the physical layout of a storm drainage system but contains placeholder conduit sizes, incomplete hydraulic parameters, and provisional elevations. The objective of this project is to develop a hydraulically consistent storm drainage design using the Rational Method for initial sizing and evaluate system performance using SWMM.

### Problem Statement Review (Engineering Objectives)

The project integrates traditional hand-based hydrologic design with hydraulic simulation. Students will:

- Delineate drainage areas contributing to each inlet using aerial imagery and elevation data.

- Estimate design flows using the Rational Method.

- Size storm sewer conduits to convey design flows for a 5-year design storm in Harris County, Texas.

- Establish the vertical geometry of the drainage system through appropriate invert elevations and pipe offsets.

- Implement the resulting design in a SWMM hydraulic model to evaluate system behavior during a design storm event.

### Design and Modeling Tasks

Students will complete the following major tasks:

1. **Watershed Delineation**

  - Construct a contour map from provided elevation data.

  - Identify drainage directions and contributing areas.

  - Assign drainage areas to inlet nodes and identify junctions needed with and without direct inflow.

2. **Hydrologic Design**

  - Compute inlet times and pipe travel times.

  - Apply the Rational Method to determine design flows at each inlet.

3. **Storm Sewer Design**

  - Size conduits to convey the calculated design flows.

  - Define system vertical geometry including:

  - invert elevations at nodes

  - conduit slopes

  - offsets and crown elevations to maintain hydraulic continuity.

4. **Hydraulic Modeling**

  - Build a SWMM model representing the designed system.

  - Define subcatchments and drainage parameters.

  - Adjust subcatchment “virtual widths” so SWMM runoff timing approximates Rational Method inlet timing.

  - Simulate a 3-hour, 5-year design storm with the outlet modeled as a free outfall to a roadside ditch.

5. **Evaluation of System Performance**

Model results will be used to assess system behavior under the design storm, including:

  - Hydraulic capacity of storm sewer conduits

  - Potential surcharge or flow limitations

  - Consistency between Rational Method design expectations and SWMM simulation results
