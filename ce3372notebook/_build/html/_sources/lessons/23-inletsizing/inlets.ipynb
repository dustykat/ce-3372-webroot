## Storm Sewers

Storm sewers are an essential component of urban drainage systems, designed to manage the runoff generated by rainfall events. Unlike sanitary sewers, which carry wastewater, storm sewers convey rainwater and surface runoff away from streets, parking lots, and other impervious surfaces to prevent flooding and waterlogging in urban areas. These systems typically consist of a **network of inlets, underground pipes, manholes, and outfalls** that discharge water into nearby water bodies such as rivers, lakes, or retention basins. Properly designed storm sewer systems help reduce flood risks, protect infrastructure, and improve safety by minimizing surface water accumulation during storms.

### Stormwater Inlet Design Considerations

Stormwater inlets are hydraulic control structures that allow surface runoff to enter the storm sewer system. Their primary purpose is to intercept and convey runoff from streets, parking areas, and other paved surfaces before excessive ponding occurs. By removing water from the roadway surface, inlets help maintain traffic safety, reduce splash and spray, and limit localized flooding that could otherwise affect adjacent infrastructure or private property.

In many urban drainage systems, inlets are integrated into curb-and-gutter street sections. In these configurations, the curb and gutter system performs multiple roles. The gutter functions as a shallow open channel that conveys runoff along the street, while the curb helps direct flow toward designated interception points. Inlets placed along the gutter line capture a portion or all of this runoff and route it through a storm sewer network. Because the curb and gutter assembly also contributes to roadway structure and traffic control, inlet placement and geometry must be compatible with these functions.

Several inlet geometries are commonly used in stormwater systems:
- Curb inlets – openings in the vertical curb face that allow water flowing in the gutter to enter the storm sewer.
- Grate (drop) inlets – horizontal grates located in the pavement surface that allow runoff to fall directly into an inlet structure below.
- Slot inlets – narrow continuous openings that intercept flow along a linear slot.
- Combination inlets – systems that combine curb openings and grates to improve interception efficiency under varying flow conditions.

Each inlet type has advantages and limitations related to hydraulic capacity, debris tolerance, safety, and maintenance. For example, curb inlets tend to be less susceptible to clogging by debris but are less efficient at capturing high gutter flows, whereas grate inlets can intercept larger flows but may be more vulnerable to blockage or bicycle hazards depending on grate design.

Practical structural constraints also influence inlet design. Curb inlets are typically supported by a subsurface inlet box or vault structure, similar to a small manhole, that connects to the storm sewer pipe. Because the curb opening must be supported by the underlying structure, curb inlet length is usually limited to a finite span (often on the order of several meters or roughly 20–25 feet). When greater interception capacity is required, designers often use multiple inlet boxes in tandem, creating a longer effective opening. Tandem installations require careful grading and hydraulic evaluation to ensure that upstream openings capture flow efficiently and do not allow excessive bypass to downstream locations.

The sizing and spacing of inlets is determined by hydraulic design considerations such as the expected runoff rate, allowable spread of water on the roadway, and the interception efficiency of the inlet type. Designers must balance several competing factors, including:

- the amount of runoff that must be captured,
- physical and structural limitations of the inlet geometry,
- allowable street ponding or gutter spread,
- debris and maintenance considerations,
- and safety requirements for vehicles, bicycles, and pedestrians.

For example, curb openings are commonly limited in vertical dimension (often on the order of 150–200 mm or about 6–8 inches), which constrains their capture capacity during deeper gutter flows. As a result, designers frequently use hydraulic relationships relating gutter flow depth, approach velocity, and inlet length to determine how much runoff can be intercepted at a given location.

Design guidance for stormwater inlets is typically provided in state transportation agency drainage manuals and national guidance documents. In the United States, commonly referenced resources include the Federal Highway Administration (FHWA) Urban Drainage Design Manual (Hydraulic Engineering Circular No. 22) and state-specific drainage design manuals (such as those published by transportation departments). These documents provide empirical and theoretical methods for estimating inlet interception efficiency, spacing requirements, and allowable roadway ponding depths.
